# Q1. Sum of Multiple Numbers using *args

## Problem Statement
Write a Python function using *args to calculate the sum of all numbers passed as arguments.

## Sample Function Call
```python
addition(10, 20, 30)
```

## Sample Output
```python
60
```


````python
def addition(*args):
    return sum(args)

# Sample Function Call
print(addition(10, 20, 30))  # Output: 60
````

# Q3. Print Student Details using **kwargs

## Problem Statement
Write a Python function using **kwargs to print student information.

## Sample Function Call
```python
student(name='Omprakash', age=22, course='Python')
```

## Sample Output
```python
name : Omprakash
age : 22
course : Python
```


````python
def student(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}", end=" ")

# Sample Function Call
student(name='Omprakash', age=22, course='Python')
# Output: name : Omprakash age : 22 course : Python
````

# Q9. Multiply All Numbers using *args

## Problem Statement
Write a Python function using *args to multiply all numbers.

## Sample Function Call
```python
multiply(2, 3, 4)
```

## Sample Output
```python
24
```


````python
def multiply(*args):
    result = 1
    for num in args:
        result *= num
    return result

# Sample Function Call
print(multiply(2, 3, 4))
````
