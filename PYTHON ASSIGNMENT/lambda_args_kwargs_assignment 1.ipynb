## 1. Print All Values using *args

### Problem Statement
Write a function using *args to print all values passed to the function.

### Example Input
```python
10, 20, 30
```

### Example Output
```python
10 20 30
```


````python
def print_all(*args):
    for val in args:
        print(val, end=" ")


# Example usage:
print_all(10, 20, 30)
````


## 2. Sum of Numbers using *args


````python
def count_args(*args):
    return len(args)


# Example usage:
print(count_args("a", "b", "c"))
````


## 4. Print Student Details using **kwargs

### Problem Statement
Write a function using **kwargs to print student details.

### Example Input
```python
name='Parth', age=22
```

### Example Output
```python
name : Parht
age : 22
```


````python
def print_student_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}", end=" ")


# Example usage:
print_student_details(name="Parth", age=22)
````


## 5. Print Keys using **kwargs

### Problem Statement
Write a function using **kwargs to print all keys.

### Example Input
```python

```

### Example Output
```python

state
```


````python
def print_keys(**kwargs):
    for key in kwargs.keys():
        print(key, end=" ")


# Example usage:
print_keys(city="Bhopal", state="MP")
````
