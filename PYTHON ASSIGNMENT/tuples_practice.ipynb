###  Question 2
Create a tuple of 3 fruits and print each element.

**Expected Output:**
```
apple
banana
mango
```

````python
fruits = ("apple", "banana", "mango")
for fruit in fruits:
    print(fruit, end=" ")
````

###  Question 6
Create a tuple and check if an element **exists** in it.

**Input:** `(10, 20, 30)`, check `20`  
**Expected Output:** `Present`

````python
t = (10, 20, 30)
if 20 in t:
    print("Present")
````

### Question 7
Create a tuple and print all elements using a **loop**.

**Input:** `(5, 6, 7)`  
**Expected Output:**
```
5
6
7
```

````python
t = (5, 6, 7)
for item in t:
    print(item, end=" ")
````

### Question 8
Create a tuple and **count** how many times an element appears.

**Input:** `(1, 2, 2, 3)`, check `2`  
**Expected Output:** `Count = 2`

````python
t = (1, 2, 2, 3)
print("Count =", t.count(2))
````
