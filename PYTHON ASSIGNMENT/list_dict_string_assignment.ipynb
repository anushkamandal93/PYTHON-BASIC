# Q1. Reverse Each Word in a String

## Problem Statement
Write a Python program to take a sentence from the user and reverse each word individually.

## Sample Input
```python
Enter sentence: hello world
```

## Sample Output
```python
olleh dlrow
```


````python
# Take sentence input from the user
sentence = input("Enter sentence: ")

# Split into words, reverse each word, and join them back
reversed_words = [word[::-1] for word in sentence.split()]
result = " ".join(reversed_words)

print(result)
````

# Q2. Find Frequency of Characters

## Problem Statement
Write a Python program to count the frequency of each character in a string.

## Sample Input
```python
Enter string: apple
```

## Sample Output
```python
a : 1
p : 2
l : 1
e : 1
```


````python
# Take string input from the user
string = input("Enter string: ")

# Count frequency of each character using a dictionary
freq = {}
for char in string:
    freq[char] = freq.get(char, 0) + 1

# Print the result
for char, count in freq.items():
    print(f"{char} : {count}")
````

# Q3. Remove Duplicate Elements from List

## Problem Statement

## Sample Input
```python
Enter list: [1,2,2,3,4,4,5]
```

## Sample Output
```python
[1, 2, 3, 4, 5]
```

# Q4. Find Second Largest Number in List

## Problem Statement
Write a Python program to find the second largest element in a list.

## Sample Input
```python
Enter list: [10, 20, 5, 40, 30]
```

## Sample Output
```python
30
```


````python
# Take list input
lst = [10, 20, 5, 40, 30]

# Remove duplicates and sort the list
unique_sorted = sorted(list(set(lst)))

# Find the second largest (second to last element)
second_largest = unique_sorted[-2]

print(second_largest)
````

# Q5. Merge Two Dictionaries

Write a Python program to merge two dictionaries into one.

```python
dict1 = {'a':1, 'b':2}

```

# Q7. Word Counter Using Dictionary

## Problem Statement
Write a Python program to count occurrence of each word in a sentence using dictionary.

## Sample Input
```python
Enter sentence: python is easy python is powerful
```

## Sample Output
```python
python : 2
is : 2
easy : 1
powerful : 1
```


````python
# Take sentence input from the user
sentence = input("Enter sentence: ")

# Count occurrence of each word using a dictionary
word_count = {}
for word in sentence.split():
    word_count[word] = word_count.get(word, 0) + 1

# Print the result
for word, count in word_count.items():
    print(f"{word} : {count}")
````

# Q8. Sort List Without Using sort()

## Problem Statement
Write a Python program to sort a list without using the built-in sort() function.

## Sample Input
```python
Enter list: [5,2,8,1]
```

## Sample Output
```python
[1,2,5,8]
```


````python
lst = [5, 2, 8, 1]

# Bubble sort algorithm without using sort()
n = len(lst)
for i in range(n):
    for j in range(0, n - i - 1):
        if lst[j] > lst[j + 1]:
            lst[j], lst[j + 1] = lst[j + 1], lst[j]

print(lst)
````

# Q9. Create Dictionary from Two Lists

## Problem Statement
Write a Python program to create a dictionary using two lists.

## Sample Input
```python
keys = ['name', 'age']
values = ['Shiva', 22]
```

## Sample Output
```python
{'name': 'Shiva', 'age': 22}
```


````python
keys = ['name', 'age']
values = ['Shiva', 22]

# Create dictionary using zip
result_dict = dict(zip(keys, values))

print(result_dict)
````

# Q10. Find Longest Word in String

## Problem Statement
Write a Python program to find the longest word in a sentence.

## Sample Input
```python
Enter sentence: Python programming is amazing
```

## Sample Output
```python
programming
```


````python
# Take sentence input from the user
sentence = input("Enter sentence: ")

# Find the longest word using max() with key=len
words = sentence.split()
longest_word = max(words, key=len)

print(longest_word)
````
