## 1. Reverse a List Without Using reverse()

### Problem Statement
Write a program to reverse a list without using the built-in reverse() method.

### Example Input
```python
1 2 3 4 5
```

### Example Output
```python
[5, 4, 3, 2, 1]
```


````python
def reverse_list(lst):
    # Using slicing with a step of -1
    return lst[::-1]


# Example usage:
my_list = [1, 2, 3, 4, 5]
print(reverse_list(my_list))
````


## 2. Find Duplicate Elements in a List

### Problem Statement
Write a program to find all duplicate elements in a list.

### Example Input
```python
1 2 3 2 4 5 1
```

### Example Output
```python
[1, 2]
```


````python
def find_duplicates(lst):
    seen = set()
    duplicates = set()
    for item in lst:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)
    return list(duplicates)


# Example usage:
my_list = [1, 2, 3, 2, 4, 5, 1]
print(find_duplicates(my_list))
````


## 3. Count Frequency of Characters

### Problem Statement
Write a program to count the frequency of each character in a string.

### Example Input
```python
hello
```

### Example Output
```python
{'h':1, 'e':1, 'l':2, 'o':1}
```


````python
def char_frequency(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq


# Example usage:
text = "hello"
print(char_frequency(text))
````


## 4. Check Palindrome String

### Problem Statement
Write a program to check whether a string is palindrome or not.

### Example Input
```python
madam
```

### Example Output
```python
Palindrome
```


````python
def is_palindrome(s):
    # Remove any spaces and convert to lowercase for accuracy
    cleaned = s.replace(" ", "").lower()
    if cleaned == cleaned[::-1]:
        return "Palindrome"
    return "Not Palindrome"


# Example usage:
print(is_palindrome("madam"))
````


## 5. Merge Two Dictionaries

### Problem Statement
Write a program to merge two dictionaries into one.

### Example Input
```python
dict1={'a':1}, dict2={'b':2}
```

### Example Output
```python
{'a':1, 'b':2}
```


````python
def merge_dictionaries(dict1, dict2):
    # Using the merge operator (|) introduced in Python 3.9
    return dict1 | dict2


# Example usage:
d1 = {"a": 1}
d2 = {"b": 2}
print(merge_dictionaries(d1, d2))
````


## 6. Find Maximum Value in Dictionary

### Problem Statement
Write a program to find the key with maximum value in a dictionary.

### Example Input
```python
{'a':10, 'b':50, 'c':20}
```

### Example Output
```python
b
```


````python
def max_value_key(d):
    # Find the key corresponding to the maximum value
    return max(d, key=d.get)


# Example usage:
my_dict = {"a": 10, "b": 50, "c": 20}
print(max_value_key(my_dict))
````


## 7. Remove Vowels from String

### Problem Statement
Write a program to remove all vowels from a string.

### Example Input
```python
education
```

### Example Output
```python
dctn
```


````python
def remove_vowels(s):
    vowels = "aeiouAEIOU"
    return "".join([char for char in s if char not in vowels])


# Example usage:
print(remove_vowels("education"))
````


## 8. Sort Words Alphabetically

### Problem Statement
Write a program to sort words in a sentence alphabetically.

### Example Input
```python
python is easy
```

### Example Output
```python
easy is python
```


````python
def sort_words(sentence):
    words = sentence.split()
    words.sort()
    return " ".join(words)


# Example usage:
print(sort_words("python is easy"))
````


## 9. Find Common Elements Between Two Lists

### Problem Statement
Write a program to find common elements between two lists.

### Example Input
```python
List1=[1,2,3,4], List2=[3,4,5,6]
```

### Example Output
```python
[3, 4]
```


````python
def common_elements(list1, list2):
    return list(set(list1) & set(list2))


# Example usage:
l1 = [1, 2, 3, 4]
l2 = [3, 4, 5, 6]
print(common_elements(l1, l2))
````


## 10. Convert List into Dictionary

### Problem Statement
Write a program to convert two lists into a dictionary.

### Example Input
```python
keys=['a','b'], values=[1,2]
```

### Example Output
```python
{'a':1, 'b':2}
```


````python
def lists_to_dict(keys, values):
    return dict(zip(keys, values))


# Example usage:
keys = ["a", "b"]
values = [1, 2]
print(lists_to_dict(keys, values))
````


## 11. Find Longest Word in a String

### Problem Statement
Write a program to find the longest word in a sentence.

### Example Input
```python
Python programming is powerful
```

### Example Output
```python
programming
```


````python
def longest_word(sentence):
    words = sentence.split()
    return max(words, key=len)


# Example usage:
print(longest_word("Python programming is powerful"))
````


## 12. Group Elements by Frequency

### Problem Statement
Write a program to group elements based on their frequency in a list.

### Example Input
```python
[1,1,2,3,3,3]
```

### Example Output
```python
{1:2, 2:1, 3:3}
```


````python
def group_by_frequency(lst):
    freq = {}
    for item in lst:
        freq[item] = freq.get(item, 0) + 1
    return freq


# Example usage:
print(group_by_frequency([1, 1, 2, 3, 3, 3]))
````


## 14. Nested Dictionary Access

### Problem Statement
Write a program to access values from a nested dictionary.

### Example Input
```python
{'student': {'name':'Vikas', 'age':22}}
```

### Example Output
```python
Vikas
```


````python
def get_nested_value(d, keys):
    for key in keys:
        d = d[key]
    return d


# Example usage:
data = {"student": {"name": "Vikas", "age": 22}}
print(get_nested_value(data, ["student", "name"]))
````


## 15. Rotate List by K Positions

### Problem Statement
Write a program to rotate a list by K positions.

### Example Input
```python
List=[1,2,3,4,5], K=2
```

### Example Output
```python
[4, 5, 1, 2, 3]
```


````python
def rotate_list(lst, k):
    k = k % len(lst)  # Handle cases where k is greater than list length
    return lst[-k:] + lst[:-k]


# Example usage:
my_list = [1, 2, 3, 4, 5]
k = 2
print(rotate_list(my_list, k))
````
