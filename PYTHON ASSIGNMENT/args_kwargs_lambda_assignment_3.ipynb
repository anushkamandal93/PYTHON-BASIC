# Q1. Find Minimum Number using *args

## Problem Statement
Write a Python function using *args to find the minimum number from given arguments.

## Sample Function Call
```python
minimum(10, 5, 30, 2, 8)
```

## Sample Output
```python
2
```


````python
def minimum(*args):
  # Check if arguments are provided to avoid ValueError
  if not args:
    return None
  return min(args)


# Sample Function Call
print(minimum(10, 5, 30, 2, 8))
````

# Q2. Sum of Even Numbers using *args

## Problem Statement
Write a Python function using *args to calculate the sum of all even numbers.

## Sample Function Call
```python
even_sum(1,2,3,4,5,6)
```

## Sample Output
```python
12
```


````python
def even_sum(*args):
  # Iterate through args, filter even numbers, and sum them up
  return sum(num for num in args if num % 2 == 0)


# Sample Function Call
print(even_sum(1, 2, 3, 4, 5, 6))
````

# Q3. User Profile Generator using **kwargs

## Problem Statement
Write a Python function using **kwargs to print a user profile dynamically.

## Sample Function Call
```python
profile(name='Aman', city='Delhi', skill='Python')
```

## Sample Output
```python
name : Aman
city : Delhi
skill : Python
```


````python
def profile(**kwargs):
  # Iterate through key-value pairs and print them in the specified format
  for key, value in kwargs.items():
    print(f"{key} : {value}")


# Sample Function Call
profile(name="Aman", city="Delhi", skill="Python")
````

# Q4. Count Keyword Arguments using **kwargs

## Problem Statement
Write a Python function using **kwargs to count total keyword arguments passed.

## Sample Function Call
```python
details(name='Arjun', age=25, course='DS')
```

## Sample Output
```python
3
```


````python
def details(**kwargs):
  # Return the length of the kwargs dictionary, which represents the count of arguments
  return len(kwargs)


# Sample Function Call
print(details(name="Arjun", age=25, course="DS"))
````

# Q6. Sort Dictionary by Value using Lambda

## Problem Statement
Write a Python program to sort a dictionary based on values using lambda.

## Sample Input
```python
{'a':3, 'b':1, 'c':2}
```

## Sample Output
```python
[('b',1), ('c',2), ('a',3)]
```


````python
# Sample Input Dictionary
data = {"a": 3, "b": 1, "c": 2}

# Sort dictionary items based on their values (item[1]) using lambda
sorted_dict = sorted(data.items(), key=lambda item: item[1])

print(sorted_dict)
````

# Q7. Filter Odd Numbers using Lambda

## Problem Statement
Write a Python program using lambda and filter() to extract odd numbers from a list.

## Sample Input
```python
[1,2,3,4,5,6,7]
```

## Sample Output
```python
[1,3,5,7]
```


````python
# Sample Input List
numbers = [1, 2, 3, 4, 5, 6, 7]

# Use filter with lambda to keep only odd numbers
odd_numbers = list(filter(lambda x: x % 2 != 0, numbers))

print(odd_numbers)
````

# Q8. Convert Strings to Uppercase using Lambda

## Problem Statement
Write a Python program using map() and lambda to convert all strings into uppercase.

## Sample Input
```python
['python', 'data', 'science']
```

## Sample Output
```python
['PYTHON', 'DATA', 'SCIENCE']
```


````python
# Sample Input List
words = ["python", "data", "science"]

# Use map with lambda to convert each string to uppercase
upper_words = list(map(lambda s: s.upper(), words))

print(upper_words)
````

# Q9. Student Marks Analyzer using *args

## Problem Statement
Write a Python function using *args to calculate:
- Total Marks
- Average Marks
- Highest Marks

## Sample Function Call
```python
marks(70,80,90,60)
```

## Sample Output
```python
Total: 300
Average: 75.0
Highest: 90
```


````python
def marks(*args):
  if not args:
    return "No marks provided."

  total = sum(args)
  average = total / len(args)
  highest = max(args)

  print(f"Total : {total}")
  print(f"Average : {average}")
  print(f"Highest : {highest}")


# Sample Function Call
marks(70, 80, 90, 60)
````

# Q10. Dynamic Bill Generator using **kwargs

## Problem Statement
Write a Python function using **kwargs to print item names and prices, and calculate total bill amount.

## Sample Function Call
```python
bill(Shirt=1200, Shoes=2500, Watch=3000)
```

## Sample Output
```python
Shirt : 1200
Shoes : 2500
Watch : 3000
Total Bill : 6700
```


````python
def bill(**kwargs):
  total_bill = 0
  for item, price in kwargs.items():
    print(f"{item} : {price}")
    total_bill += price
  print(f"Total Bill : {total_bill}")


# Sample Function Call
bill(Shirt=1200, Shoes=2500, Watch=3000)
````
